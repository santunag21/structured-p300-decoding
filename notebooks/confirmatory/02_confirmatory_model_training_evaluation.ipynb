```python
# ============================================================
# A01 FOLD 6 — SAFE NEURAL TRAINING RESUME
#
# PURPOSE
#   Recover ONLY missing Fold-6 neural fits.
#
# EXPECTED missing fits from the previous audit:
#   seed 1: binary_dbnet
#   seed 2: binary_dbnet
#   seed 3: structured_dbnet
#           glass_gated_dbnet
#           interaction_dbnet
#           binary_dbnet
#
# IMPORTANT
#   - Existing completed fits are NEVER retrained.
#   - Existing Fold-6 GLASS Stage-2 betaMat is REUSED.
#   - Outer Fold-6 EEG/labels are NOT constructed or evaluated here.
#   - Any unexpected partial run is archived before retraining.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import gc
import hashlib
import importlib.util
import json
import sys
import time

import numpy as np
import tensorflow as tf


# ============================================================
# 0. PATHS
# ============================================================

PROJECT_ROOT = Path(
    "/content/drive/.shortcut-targets-by-id/"
    "1un2SaLv7_DvXerUlPtzKSNZ_j-mu1qFQ/"
    "EEG_GANet_Reproduction"
)

CV_ROOT = (
    PROJECT_ROOT
    / "GLASS_GANet/results/paper_confirmatory_cv_v1"
)

PREPARED_DIR = PROJECT_ROOT / "GLASS_GANet/prepared_data"

SUBJECT = "A01"
FOLD = 6

MODEL_ROOT = (
    CV_ROOT
    / "models/automated_cv"
    / SUBJECT
    / f"outer_fold_{FOLD}"
)

GLASS_STAGE2 = (
    CV_ROOT
    / "models/pure_glass"
    / SUBJECT
    / f"outer_fold_{FOLD}"
    / "stage2_shrinkage.npz"
)

CANONICAL_DATA = (
    PREPARED_DIR
    / "A01_GLASS_preprocessed_seed42.npz"
)


# ============================================================
# 1. FROZEN SETTINGS
# ============================================================

LEARNING_RATE = 0.0009

MAXIMUM_EPOCHS = 1000
PATIENCE = 50
MINIMUM_DELTA = 1e-4

GROUP_BATCH_SIZE = 32
BINARY_BATCH_SIZE = 64

SEEDS = [1, 2, 3]

EXPECTED_PARAMETER_COUNTS = {
    "structured_dbnet": 3961,
    "glass_gated_dbnet": 4986,
    "interaction_dbnet": 3989,
    "binary_dbnet": 4090,
}

# Frozen Fold-6 TRAIN + INNER VALIDATION only.
# Outer-test identities are intentionally not needed by this cell.
TRAIN_CHARACTERS = [
    2, 4, 5, 6, 7, 8, 9, 10, 12,
    15, 16, 17, 18, 19,
    22, 23, 24, 25, 26, 27,
    29, 31, 32, 33, 35
]

VALIDATION_CHARACTERS = [
    1, 3, 13, 14, 30
]


# ============================================================
# 2. BASIC AUDIT
# ============================================================

def require(condition, message):
    if not condition:
        raise RuntimeError(message)


def sha256_file(path):
    digest = hashlib.sha256()
    with Path(path).open("rb") as handle:
        for block in iter(
            lambda: handle.read(1024 * 1024),
            b""
        ):
            digest.update(block)
    return digest.hexdigest()


require(PROJECT_ROOT.is_dir(), f"Missing project:\n{PROJECT_ROOT}")
require(CV_ROOT.is_dir(), f"Missing CV root:\n{CV_ROOT}")
require(CANONICAL_DATA.is_file(), f"Missing dataset:\n{CANONICAL_DATA}")
require(GLASS_STAGE2.is_file(), f"Missing Fold-6 GLASS fit:\n{GLASS_STAGE2}")

gpu = tf.config.list_physical_devices("GPU")
require(gpu, "No GPU visible.")

try:
    tf.config.experimental.enable_op_determinism()
    print("Deterministic TensorFlow ops: ENABLED")
except Exception as error:
    print("Deterministic-op request:", error)

print("TensorFlow:", tf.__version__)
print("GPU:", gpu)


# ============================================================
# 3. IMPORT PINNED DBNET IMPLEMENTATION
# ============================================================

CODE_DIR = PROJECT_ROOT / "code/EEG-GANet"
MODEL_SOURCE = CODE_DIR / "github_model.py"

require(MODEL_SOURCE.is_file(), f"Missing:\n{MODEL_SOURCE}")

# Previously certified public source hash.
EXPECTED_DBNET_SHA256 = (
    "f3e570d12de34080ea13003a37bf9a2493b93f1a2376631e8224a40942231cd5"
)

actual_dbnet_hash = sha256_file(MODEL_SOURCE)

require(
    actual_dbnet_hash == EXPECTED_DBNET_SHA256,
    "DBNet source hash changed.\n"
    f"Expected: {EXPECTED_DBNET_SHA256}\n"
    f"Actual:   {actual_dbnet_hash}"
)

if str(CODE_DIR) not in sys.path:
    sys.path.insert(0, str(CODE_DIR))

spec = importlib.util.spec_from_file_location(
    "github_model_fold6_resume",
    MODEL_SOURCE
)
module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(module)

EEG_DBNet_V2 = module.EEG_DBNet_V2

print("DBNet source hash: PASS")


# ============================================================
# 4. LOAD CANONICAL FULL A01 DATA
#    TRAIN + INNER VALIDATION ARE CONSTRUCTED.
#    OUTER TEST IS NOT CONSTRUCTED.
# ============================================================

with np.load(CANONICAL_DATA, allow_pickle=False) as archive:

    X_all = np.asarray(
        archive["X_glass_processed_grouped"],
        dtype=np.float32
    )

    y_all = np.asarray(
        archive["y_grouped"],
        dtype=np.int8
    )

require(
    X_all.shape == (35, 10, 2, 6, 8, 128),
    f"Unexpected EEG shape: {X_all.shape}"
)

require(
    y_all.shape == (35, 10, 2, 6),
    f"Unexpected label shape: {y_all.shape}"
)

require(np.isfinite(X_all).all(), "Non-finite EEG.")
require(np.all(y_all.sum(axis=-1) == 1), "Invalid six-choice labels.")


def chars_to_zero_based(character_ids):
    ids = np.asarray(character_ids, dtype=np.int64)
    require(
        np.all((ids >= 1) & (ids <= 35)),
        "Character IDs out of range."
    )
    return ids - 1


train_idx = chars_to_zero_based(TRAIN_CHARACTERS)
val_idx = chars_to_zero_based(VALIDATION_CHARACTERS)

require(
    np.intersect1d(train_idx, val_idx).size == 0,
    "Train/validation character overlap."
)

X_train = (
    X_all[train_idx]
    .reshape(-1, 6, 8, 128)
    [..., np.newaxis]
)

y_train = (
    y_all[train_idx]
    .reshape(-1, 6)
    .astype(np.float32)
)

X_val = (
    X_all[val_idx]
    .reshape(-1, 6, 8, 128)
    [..., np.newaxis]
)

y_val = (
    y_all[val_idx]
    .reshape(-1, 6)
    .astype(np.float32)
)

require(X_train.shape == (500, 6, 8, 128, 1), X_train.shape)
require(y_train.shape == (500, 6), y_train.shape)

require(X_val.shape == (100, 6, 8, 128, 1), X_val.shape)
require(y_val.shape == (100, 6), y_val.shape)

print("\nFold-6 training data: PASS")
print("Structured train:", X_train.shape, y_train.shape)
print("Structured val:  ", X_val.shape, y_val.shape)
print("Outer performance evaluated: NO")


# ============================================================
# 5. REUSE EXISTING FOLD-LOCAL GLASS betaMat
# ============================================================

with np.load(GLASS_STAGE2, allow_pickle=False) as archive:

    print("\nFold-6 GLASS Stage-2 keys:")
    print(archive.files)

    require(
        "betaMat" in archive.files,
        "stage2_shrinkage.npz does not contain betaMat."
    )

    beta_fold6 = np.asarray(
        archive["betaMat"],
        dtype=np.float32
    )

require(beta_fold6.shape == (8, 128), beta_fold6.shape)
require(np.isfinite(beta_fold6).all(), "Invalid betaMat.")

print("Fold-local GLASS betaMat:", beta_fold6.shape)
print("Existing GLASS fit reused: YES")
print("GLASS retraining performed: NO")


# ============================================================
# 6. EXACT STRUCTURED DBNET
# ============================================================

def build_epoch_scorer():

    public_binary_model = EEG_DBNet_V2(
        NumFilter=8,
        SamplingFrequency=128,
        NumChannels=8,
        FilterScaler=2,
        NumClasses=2,
        DropoutRate=0.5,
    ).build_model()

    require(
        isinstance(
            public_binary_model.layers[-3],
            tf.keras.layers.Concatenate
        ),
        "Unexpected public DBNet layer layout."
    )

    feature_extractor = tf.keras.Model(
        inputs=public_binary_model.input,
        outputs=public_binary_model.layers[-3].output,
        name="dbnet_v2_feature_extractor",
    )

    epoch_input = tf.keras.Input(
        shape=(8, 128, 1),
        name="single_stimulus_epoch"
    )

    features = feature_extractor(epoch_input)

    score = tf.keras.layers.Dense(
        1,
        kernel_constraint=tf.keras.constraints.max_norm(0.25),
        name="stimulus_score",
    )(features)

    return tf.keras.Model(
        epoch_input,
        score,
        name="dbnet_v2_epoch_scorer"
    )


def build_structured_dbnet():

    epoch_scorer = build_epoch_scorer()

    group_input = tf.keras.Input(
        shape=(6, 8, 128, 1),
        name="six_stimulus_group"
    )

    scores = tf.keras.layers.TimeDistributed(
        epoch_scorer,
        name="shared_epoch_scorer"
    )(group_input)

    logits = tf.keras.layers.Reshape(
        (6,),
        name="six_choice_logits"
    )(scores)

    probabilities = tf.keras.layers.Softmax(
        axis=-1,
        name="six_choice_probabilities"
    )(logits)

    model = tf.keras.Model(
        group_input,
        probabilities,
        name="structured_dbnet_v2"
    )

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=LEARNING_RATE
        ),
        loss=tf.keras.losses.CategoricalCrossentropy(),
        metrics=[
            tf.keras.metrics.CategoricalAccuracy(
                name="six_choice_accuracy"
            )
        ],
    )

    return model


# ============================================================
# 7. EXACT GLASS-GATED DBNET
# ============================================================

class FixedGlassScores(tf.keras.layers.Layer):

    def __init__(self, beta_matrix, **kwargs):

        super().__init__(
            trainable=False,
            **kwargs
        )

        beta_matrix = np.asarray(
            beta_matrix,
            dtype=np.float32
        )

        require(
            beta_matrix.shape == (8, 128),
            f"Bad GLASS beta shape {beta_matrix.shape}"
        )

        self._initial_beta = beta_matrix


    def build(self, input_shape):

        self.beta_matrix = self.add_weight(
            name="beta_matrix",
            shape=(8, 128),
            initializer=tf.keras.initializers.Constant(
                self._initial_beta
            ),
            trainable=False,
        )

        super().build(input_shape)


    def call(self, inputs):

        return tf.einsum(
            "bkct,ct->bk",
            tf.cast(
                inputs[..., 0],
                self.compute_dtype
            ),
            self.beta_matrix,
        )


class NonnegativeGlassResidual(tf.keras.layers.Layer):

    def build(self, input_shape):

        self.glass_gate = self.add_weight(
            name="glass_gate",
            shape=(),
            initializer="zeros",
            trainable=True,
            constraint=tf.keras.constraints.NonNeg(),
        )

        super().build(input_shape)


    def call(self, inputs):

        dbnet_logits, glass_logits = inputs

        centered = (
            glass_logits
            -
            tf.reduce_mean(
                glass_logits,
                axis=-1,
                keepdims=True
            )
        )

        scale = tf.math.reduce_std(
            centered,
            axis=-1,
            keepdims=True
        )

        normalized_glass = (
            centered
            /
            (scale + 1e-6)
        )

        return (
            dbnet_logits
            +
            self.glass_gate * normalized_glass
        )


def build_gated_hybrid(beta_matrix):

    epoch_scorer = build_epoch_scorer()

    group_input = tf.keras.Input(
        shape=(6, 8, 128, 1),
        name="six_stimulus_group"
    )

    dbnet_scores = tf.keras.layers.TimeDistributed(
        epoch_scorer,
        name="shared_dbnet_scorer"
    )(group_input)

    dbnet_logits = tf.keras.layers.Reshape(
        (6,),
        name="dbnet_logits"
    )(dbnet_scores)

    glass_logits = FixedGlassScores(
        beta_matrix,
        name="fixed_official_glass_scores"
    )(group_input)

    fused_logits = NonnegativeGlassResidual(
        name="nonnegative_glass_residual"
    )([
        dbnet_logits,
        glass_logits
    ])

    probabilities = tf.keras.layers.Softmax(
        axis=-1,
        name="six_choice_probabilities"
    )(fused_logits)

    model = tf.keras.Model(
        group_input,
        probabilities,
        name="glass_gated_structured_dbnet"
    )

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=LEARNING_RATE
        ),
        loss=tf.keras.losses.CategoricalCrossentropy(),
        metrics=[
            tf.keras.metrics.CategoricalAccuracy(
                name="six_choice_accuracy"
            )
        ],
    )

    return model


# ============================================================
# 8. EXACT RTGP-INSPIRED INTERACTION FEATURES/MODEL
# ============================================================

CHANNEL_PAIRS = np.asarray(
    [
        (u, v)
        for u in range(7)
        for v in range(u + 1, 8)
    ],
    dtype=np.int64
)


def compute_fisher_features(grouped_eeg):

    grouped_eeg = np.asarray(
        grouped_eeg,
        dtype=np.float64
    )

    require(
        grouped_eeg.ndim == 5,
        f"Expected 5-D grouped EEG; got {grouped_eeg.shape}"
    )

    eeg = grouped_eeg[..., 0]

    groups = eeg.shape[0]

    flat = eeg.reshape(
        groups * 6,
        8,
        128
    )

    channel_std = np.std(
        flat,
        axis=-1
    )

    require(
        np.all(channel_std > 0),
        "Constant-channel epoch detected."
    )

    features = np.empty(
        (flat.shape[0], 28),
        dtype=np.float64
    )

    for pair_index, (
        channel_a,
        channel_b
    ) in enumerate(CHANNEL_PAIRS):

        a = flat[:, channel_a, :]
        b = flat[:, channel_b, :]

        a_centered = (
            a
            -
            np.mean(
                a,
                axis=1,
                keepdims=True
            )
        )

        b_centered = (
            b
            -
            np.mean(
                b,
                axis=1,
                keepdims=True
            )
        )

        numerator = np.sum(
            a_centered * b_centered,
            axis=1
        )

        denominator = np.sqrt(
            np.sum(
                a_centered ** 2,
                axis=1
            )
            *
            np.sum(
                b_centered ** 2,
                axis=1
            )
        )

        r = numerator / denominator

        r = np.clip(
            r,
            -0.999999,
            0.999999
        )

        features[:, pair_index] = (
            np.arctanh(r)
        )

    require(
        np.isfinite(features).all(),
        "Non-finite Fisher-z features."
    )

    return (
        features
        .reshape(groups, 6, 28)
        .astype(np.float32)
    )


def build_interaction_dbnet():

    epoch_scorer = build_epoch_scorer()

    eeg_input = tf.keras.Input(
        shape=(6, 8, 128, 1),
        name="six_stimulus_eeg"
    )

    interaction_input = tf.keras.Input(
        shape=(6, 28),
        name="six_stimulus_rtgp_interactions"
    )

    dbnet_scores = tf.keras.layers.TimeDistributed(
        epoch_scorer,
        name="shared_dbnet_scorer"
    )(eeg_input)

    dbnet_logits = tf.keras.layers.Reshape(
        (6,),
        name="dbnet_logits"
    )(dbnet_scores)

    interaction_scores = (
        tf.keras.layers.TimeDistributed(
            tf.keras.layers.Dense(
                1,
                use_bias=False,
                kernel_initializer="zeros",
                name="interaction_linear"
            ),
            name="shared_rtgp_interaction_scorer"
        )(interaction_input)
    )

    interaction_logits = tf.keras.layers.Reshape(
        (6,),
        name="interaction_logits_unscaled"
    )(interaction_scores)

    interaction_logits = tf.keras.layers.Rescaling(
        scale=1.0 / 28.0,
        name="rtgp_ns2_scaling"
    )(interaction_logits)

    combined_logits = tf.keras.layers.Add(
        name="combined_choice_logits"
    )([
        dbnet_logits,
        interaction_logits
    ])

    probabilities = tf.keras.layers.Softmax(
        axis=-1,
        name="six_choice_probabilities"
    )(combined_logits)

    model = tf.keras.Model(
        [
            eeg_input,
            interaction_input
        ],
        probabilities,
        name="structured_dbnet_plus_rtgp_interactions"
    )

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=LEARNING_RATE
        ),
        loss=tf.keras.losses.CategoricalCrossentropy(),
        metrics=[
            tf.keras.metrics.CategoricalAccuracy(
                name="six_choice_accuracy"
            )
        ],
    )

    return model


print("\nComputing TRAIN Fisher-z features...")
X0_train = compute_fisher_features(X_train)

print("Computing VALIDATION Fisher-z features...")
X0_val = compute_fisher_features(X_val)

require(X0_train.shape == (500, 6, 28), X0_train.shape)
require(X0_val.shape == (100, 6, 28), X0_val.shape)


# ============================================================
# 9. EXACT BINARY DBNET
# ============================================================

class TargetAUC(tf.keras.metrics.AUC):

    def __init__(
        self,
        name="target_auc",
        **kwargs
    ):
        super().__init__(
            name=name,
            **kwargs
        )

    def update_state(
        self,
        y_true,
        y_pred,
        sample_weight=None
    ):
        return super().update_state(
            y_true[:, 1],
            y_pred[:, 1],
            sample_weight=sample_weight
        )


def build_binary_dbnet():

    model = EEG_DBNet_V2(
        NumFilter=8,
        SamplingFrequency=128,
        NumChannels=8,
        FilterScaler=2,
        NumClasses=2,
        DropoutRate=0.5,
    ).build_model()

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=LEARNING_RATE
        ),
        loss=tf.keras.losses.CategoricalCrossentropy(),
        metrics=[
            tf.keras.metrics.CategoricalAccuracy(
                name="binary_accuracy"
            ),
            TargetAUC(
                name="target_auc"
            )
        ],
    )

    return model


X_binary_train = X_train.reshape(
    -1,
    8,
    128,
    1
)

binary_y_train_int = (
    y_train
    .reshape(-1)
    .astype(np.int32)
)

y_binary_train = tf.keras.utils.to_categorical(
    binary_y_train_int,
    num_classes=2
).astype(np.float32)

X_binary_val = X_val.reshape(
    -1,
    8,
    128,
    1
)

binary_y_val_int = (
    y_val
    .reshape(-1)
    .astype(np.int32)
)

y_binary_val = tf.keras.utils.to_categorical(
    binary_y_val_int,
    num_classes=2
).astype(np.float32)

require(
    X_binary_train.shape == (3000, 8, 128, 1),
    X_binary_train.shape
)

require(
    X_binary_val.shape == (600, 8, 128, 1),
    X_binary_val.shape
)

require(
    int(binary_y_train_int.sum()) == 500,
    "Unexpected target count."
)

print("\nBinary train:", X_binary_train.shape)
print(
    "Targets/non-targets:",
    int(binary_y_train_int.sum()),
    int(len(binary_y_train_int) - binary_y_train_int.sum())
)


# ============================================================
# 10. CHECK MODEL PARAMETER COUNTS BEFORE TRAINING
# ============================================================

builders_for_audit = {
    "structured_dbnet":
        lambda: build_structured_dbnet(),

    "glass_gated_dbnet":
        lambda: build_gated_hybrid(
            beta_fold6
        ),

    "interaction_dbnet":
        lambda: build_interaction_dbnet(),

    "binary_dbnet":
        lambda: build_binary_dbnet(),
}

print("\n" + "=" * 72)
print("MODEL RECONSTRUCTION AUDIT")
print("=" * 72)

for name, builder in builders_for_audit.items():

    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(1)

    model = builder()

    actual = int(
        model.count_params()
    )

    expected = EXPECTED_PARAMETER_COUNTS[
        name
    ]

    print(
        f"{name:24s}"
        f" actual={actual}"
        f" expected={expected}"
    )

    require(
        actual == expected,
        f"{name}: parameter mismatch."
    )

    del model
    gc.collect()

print("Model reconstruction audit: PASS")


# ============================================================
# 11. COMPLETION / PARTIAL-RUN HANDLING
# ============================================================

def run_directory(seed, model_name):

    return (
        MODEL_ROOT
        / f"seed_{seed}"
        / model_name
    )


def is_completed(run_dir):

    return (
        (run_dir / "best.weights.h5").is_file()
        and
        (run_dir / "training_summary.json").is_file()
    )


def archive_partial_if_needed(
    run_dir
):

    run_dir = Path(run_dir)

    if not run_dir.exists():
        run_dir.mkdir(
            parents=True,
            exist_ok=True
        )
        return

    if is_completed(run_dir):
        return

    contents = list(
        run_dir.iterdir()
    )

    if not contents:
        return

    stamp = datetime.now(
        timezone.utc
    ).strftime(
        "%Y%m%dT%H%M%SZ"
    )

    archived = run_dir.with_name(
        run_dir.name
        +
        f"_PARTIAL_RESUME_{stamp}"
    )

    require(
        not archived.exists(),
        f"Archive path already exists:\n{archived}"
    )

    run_dir.rename(archived)

    run_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    print(
        "Archived incomplete run:",
        archived
    )


def write_json_atomic(
    data,
    path
):

    path = Path(path)

    temporary = path.with_suffix(
        path.suffix + ".temporary"
    )

    with temporary.open(
        "w",
        encoding="utf-8"
    ) as handle:

        json.dump(
            data,
            handle,
            indent=2,
            allow_nan=False
        )

    temporary.replace(path)


# ============================================================
# 12. TRAIN ONE MISSING FIT
# ============================================================

def train_one(
    seed,
    model_name
):

    run_dir = run_directory(
        seed,
        model_name
    )

    run_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    if is_completed(run_dir):

        print(
            f"SKIP complete: "
            f"seed {seed} | {model_name}"
        )

        return False

    archive_partial_if_needed(
        run_dir
    )

    tf.keras.backend.clear_session()
    gc.collect()

    tf.keras.utils.set_random_seed(
        seed
    )

    if model_name == "structured_dbnet":

        model = build_structured_dbnet()

        fit_x = X_train
        fit_y = y_train

        val_x = X_val
        val_y = y_val

        monitor = "val_loss"
        mode = "min"
        min_delta = MINIMUM_DELTA
        batch_size = GROUP_BATCH_SIZE

    elif model_name == "glass_gated_dbnet":

        model = build_gated_hybrid(
            beta_fold6
        )

        fit_x = X_train
        fit_y = y_train

        val_x = X_val
        val_y = y_val

        monitor = "val_loss"
        mode = "min"
        min_delta = MINIMUM_DELTA
        batch_size = GROUP_BATCH_SIZE

    elif model_name == "interaction_dbnet":

        model = build_interaction_dbnet()

        fit_x = [
            X_train,
            X0_train
        ]

        fit_y = y_train

        val_x = [
            X_val,
            X0_val
        ]

        val_y = y_val

        monitor = "val_loss"
        mode = "min"
        min_delta = MINIMUM_DELTA
        batch_size = GROUP_BATCH_SIZE

    elif model_name == "binary_dbnet":

        model = build_binary_dbnet()

        fit_x = X_binary_train
        fit_y = y_binary_train

        val_x = X_binary_val
        val_y = y_binary_val

        monitor = "val_target_auc"
        mode = "max"
        min_delta = 0.0
        batch_size = BINARY_BATCH_SIZE

    else:
        raise ValueError(
            model_name
        )

    require(
        int(model.count_params())
        ==
        EXPECTED_PARAMETER_COUNTS[
            model_name
        ],
        f"{model_name}: parameter mismatch."
    )

    weights_path = (
        run_dir
        / "best.weights.h5"
    )

    history_path = (
        run_dir
        / "history.csv"
    )

    callbacks = [
        tf.keras.callbacks.TerminateOnNaN(),

        tf.keras.callbacks.ModelCheckpoint(
            filepath=str(
                weights_path
            ),
            monitor=monitor,
            mode=mode,
            save_best_only=True,
            save_weights_only=True,
            verbose=0,
        ),

        tf.keras.callbacks.EarlyStopping(
            monitor=monitor,
            mode=mode,
            patience=PATIENCE,
            min_delta=min_delta,
            restore_best_weights=True,
            verbose=1,
        ),

        tf.keras.callbacks.CSVLogger(
            str(history_path),
            append=False
        ),
    ]

    print("\n" + "=" * 72)
    print(
        f"TRAINING seed {seed} | "
        f"{model_name}"
    )
    print("=" * 72)

    start = time.perf_counter()

    history = model.fit(
        fit_x,
        fit_y,
        validation_data=(
            val_x,
            val_y
        ),
        epochs=MAXIMUM_EPOCHS,
        batch_size=batch_size,
        shuffle=True,
        callbacks=callbacks,
        verbose=0,
    )

    elapsed = float(
        time.perf_counter()
        -
        start
    )

    require(
        weights_path.is_file(),
        "Best checkpoint was not saved."
    )

    # Always reload the frozen checkpoint,
    # independent of EarlyStopping's in-memory state.
    model.load_weights(
        weights_path
    )

    monitor_history = np.asarray(
        history.history[monitor],
        dtype=np.float64
    )

    require(
        monitor_history.ndim == 1
        and
        len(monitor_history) > 0,
        "Missing monitor history."
    )

    if mode == "min":
        best_index = int(
            np.nanargmin(
                monitor_history
            )
        )
    else:
        best_index = int(
            np.nanargmax(
                monitor_history
            )
        )

    best_epoch = (
        best_index + 1
    )

    best_value = float(
        monitor_history[
            best_index
        ]
    )

    epochs_run = int(
        len(
            monitor_history
        )
    )

    # Training/inner-validation provenance only.
    summary = {
        "status":
            "completed",

        "subject":
            SUBJECT,

        "outer_fold":
            FOLD,

        "seed":
            int(seed),

        "model":
            model_name,

        "best_epoch":
            int(best_epoch),

        "epochs_run":
            epochs_run,

        "checkpoint_monitor":
            monitor,

        "checkpoint_mode":
            mode,

        "best_monitor_value":
            best_value,

        "total_fit_seconds":
            elapsed,

        "seconds_per_epoch":
            elapsed
            /
            epochs_run,

        "tensorflow_version":
            tf.__version__,

        "parameter_count":
            int(
                model.count_params()
            ),

        "outer_test_loaded":
            False,

        "outer_test_evaluated":
            False,

        "completed_utc":
            datetime.now(
                timezone.utc
            ).isoformat(),
    }

    # Preserve the learned gate for provenance,
    # but it is ALSO already stored in best.weights.h5.
    if model_name == "glass_gated_dbnet":

        summary[
            "glass_gate"
        ] = float(
            model.get_layer(
                "nonnegative_glass_residual"
            ).glass_gate.numpy()
        )

    write_json_atomic(
        summary,
        run_dir
        / "training_summary.json"
    )

    print(
        f"COMPLETE | "
        f"seed {seed} | "
        f"{model_name} | "
        f"{elapsed/60:.2f} min | "
        f"best epoch {best_epoch} | "
        f"{monitor}={best_value:.6f}"
    )

    del model
    gc.collect()
    tf.keras.backend.clear_session()

    return True


# ============================================================
# 13. RESUME ALL 12 SLOTS — COMPLETED ONES ARE SKIPPED
# ============================================================

MODEL_ORDER = [
    "structured_dbnet",
    "glass_gated_dbnet",
    "interaction_dbnet",
    "binary_dbnet",
]

print("\n" + "=" * 72)
print("PRE-RESUME STATUS")
print("=" * 72)

for seed in SEEDS:
    for model_name in MODEL_ORDER:

        run_dir = run_directory(
            seed,
            model_name
        )

        state = (
            "COMPLETE"
            if is_completed(run_dir)
            else "MISSING/PARTIAL"
        )

        print(
            f"seed {seed} | "
            f"{model_name:24s} | "
            f"{state}"
        )


trained_now = []

for seed in SEEDS:

    for model_name in MODEL_ORDER:

        did_train = train_one(
            seed,
            model_name
        )

        if did_train:

            trained_now.append(
                (
                    seed,
                    model_name
                )
            )


# ============================================================
# 14. FINAL CHECK — ALL 12 FITS MUST EXIST
# ============================================================

missing = []

for seed in SEEDS:

    for model_name in MODEL_ORDER:

        run_dir = run_directory(
            seed,
            model_name
        )

        if not is_completed(
            run_dir
        ):
            missing.append(
                (
                    seed,
                    model_name
                )
            )

require(
    not missing,
    "Still missing runs: "
    + repr(missing)
)


# ============================================================
# 15. LOAD-WEIGHT COMPATIBILITY CHECK
#     STILL NO OUTER TEST ACCESS
# ============================================================

print("\n" + "=" * 72)
print("CHECKPOINT LOAD AUDIT")
print("=" * 72)

for seed in SEEDS:

    for model_name in MODEL_ORDER:

        tf.keras.backend.clear_session()
        tf.keras.utils.set_random_seed(seed)

        if model_name == "structured_dbnet":

            model = build_structured_dbnet()

        elif model_name == "glass_gated_dbnet":

            model = build_gated_hybrid(
                beta_fold6
            )

        elif model_name == "interaction_dbnet":

            model = build_interaction_dbnet()

        else:

            model = build_binary_dbnet()

        path = (
            run_directory(
                seed,
                model_name
            )
            /
            "best.weights.h5"
        )

        model.load_weights(
            path
        )

        print(
            f"PASS | seed {seed} | "
            f"{model_name}"
        )

        del model
        gc.collect()


print("\n" + "=" * 72)
print("A01 FOLD-6 TRAINING RECOVERY COMPLETE")
print("=" * 72)

print("Fits trained in this recovery:")
for seed, model_name in trained_now:
    print(
        f"  seed {seed} | {model_name}"
    )

print(
    "\nAll 12 neural checkpoints: PASS"
)

print(
    "Existing Fold-6 GLASS fit: REUSED"
)

print(
    "Outer Fold-6 EEG/labels accessed: NO"
)

print(
    "Outer Fold-6 performance evaluated: NO"
)

print(
    "\nSTOP HERE. Do not manually open the outer test yet."
)
```

```python
# ============================================================
# A01 OUTER FOLD 6 — LOCKED OUTER EVALUATION
#
# Run ONLY after:
#   "All 12 neural checkpoints: PASS"
#
# NO TRAINING
# NO MODEL SELECTION
# NO TEMPERATURE TUNING
#
# Frozen rules:
#   Neural seeds 1,2,3 -> average probabilities FIRST.
#   Binary:
#       average 2-class probabilities first
#       p = ensemble target probability
#       logit = log(p/(1-p))
#       reshape six choices
#       softmax T = 1
#
#   GLASS:
#       official median prediction using fold-local betaMat
#
# Primary:
#       GLASS-gated vs Structured at Acc@4
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import ast
import gc
import hashlib
import json
import types
import urllib.request

import numpy as np
import pandas as pd
import tensorflow as tf

from scipy.special import softmax

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
)


# ============================================================
# 1. FROZEN IDENTIFIERS
# ============================================================

SUBJECT = "A01"
FOLD = 6

SEEDS = [1, 2, 3]

OUTER_CHARACTERS = [
    11, 20, 21, 28, 34
]

EXPECTED_PARAMETER_COUNTS = {
    "structured_dbnet": 3961,
    "glass_gated_dbnet": 4986,
    "interaction_dbnet": 3989,
    "binary_dbnet": 4090,
}

OUTER_RESULT_DIR = (
    CV_ROOT
    / "outer_results"
    / SUBJECT
    / f"outer_fold_{FOLD}"
)

OUTER_PREDICTIONS_PATH = (
    OUTER_RESULT_DIR
    / "outer_predictions.npz"
)

OUTER_MANIFEST_PATH = (
    OUTER_RESULT_DIR
    / "outer_result_manifest.json"
)


def require(condition, message):
    if not condition:
        raise RuntimeError(message)


def file_sha256(path):

    h = hashlib.sha256()

    with Path(path).open("rb") as handle:
        for block in iter(
            lambda: handle.read(1024 * 1024),
            b""
        ):
            h.update(block)

    return h.hexdigest()


# ============================================================
# 2. PRE-OPEN AUDIT
# ============================================================

print("=" * 100)
print("A01 OUTER FOLD 6 — LOCKED EVALUATION")
print("=" * 100)

require(
    not OUTER_PREDICTIONS_PATH.exists(),
    "Fold 6 outer_predictions.npz already exists. "
    "Do NOT evaluate this fold twice."
)

require(
    not OUTER_MANIFEST_PATH.exists(),
    "Fold 6 outer_result_manifest.json already exists."
)


# Every neural run must be complete BEFORE opening outer data.

def neural_run_dir(
    seed,
    model_name
):

    return (
        MODEL_ROOT
        / f"seed_{seed}"
        / model_name
    )


MODEL_NAMES = [
    "structured_dbnet",
    "glass_gated_dbnet",
    "interaction_dbnet",
    "binary_dbnet",
]

for seed in SEEDS:

    for model_name in MODEL_NAMES:

        run_dir = neural_run_dir(
            seed,
            model_name
        )

        require(
            (
                run_dir
                / "best.weights.h5"
            ).is_file(),
            f"Missing checkpoint: seed {seed} {model_name}"
        )

        require(
            (
                run_dir
                / "training_summary.json"
            ).is_file(),
            f"Missing training summary: seed {seed} {model_name}"
        )


require(
    GLASS_STAGE2.is_file(),
    "Fold-local GLASS Stage-2 result missing."
)

print("All 12 neural fits: PASS")
print("Fold-local GLASS fit: PASS")

print(
    "\nOuter performance evaluation starts NOW."
)


# ============================================================
# 3. CONSTRUCT OUTER DATA — FIRST OUTER ACCESS
# ============================================================

outer_idx = (
    np.asarray(
        OUTER_CHARACTERS,
        dtype=np.int64
    )
    - 1
)

X_outer_character = np.asarray(
    X_all[outer_idx],
    dtype=np.float32
)

y_outer_character = np.asarray(
    y_all[outer_idx],
    dtype=np.int8
)

require(
    X_outer_character.shape
    ==
    (5, 10, 2, 6, 8, 128),
    X_outer_character.shape
)

require(
    y_outer_character.shape
    ==
    (5, 10, 2, 6),
    y_outer_character.shape
)


# Make certain each character's target remains identical
# across all 10 repetitions.

targets = (
    y_outer_character
    .argmax(axis=-1)
)

require(
    targets.shape == (5, 10, 2),
    targets.shape
)

np.testing.assert_array_equal(
    targets,
    np.repeat(
        targets[:, :1, :],
        repeats=10,
        axis=1
    )
)


# 100 six-choice half-sequence groups.

X_outer = (
    X_outer_character
    .reshape(
        100,
        6,
        8,
        128
    )
    [..., np.newaxis]
)

y_outer = (
    y_outer_character
    .reshape(
        100,
        6
    )
    .astype(np.float32)
)

require(
    X_outer.shape
    ==
    (100, 6, 8, 128, 1),
    X_outer.shape
)

require(
    y_outer.shape
    ==
    (100, 6),
    y_outer.shape
)

require(
    np.all(
        y_outer.sum(axis=1)
        == 1
    ),
    "Invalid outer labels."
)

print(
    "Outer EEG:",
    X_outer.shape
)

print(
    "Outer labels:",
    y_outer.shape
)


# ============================================================
# 4. PINNED OFFICIAL GLASS FUNCTIONS
# ============================================================

GLASS_COMMIT = (
    "db32c2abd9211dec225a7f39a33963f8a0b21d7f"
)

PUBLIC_HASHES = {
    "code/glass.py":
        "00de59d0ba107a7a5770565bdfb5bf7dfbb372f92bed2c4aecd1bb21deda652d",

    "code/helper_funcs.py":
        "f9cf4a5134acbbda7fcb55e6503096f1698b0f9e63e16a2a4f9252dd51eb54d9",
}


def bytes_sha256(raw):

    return hashlib.sha256(
        raw
    ).hexdigest()


def obtain_pinned_source(
    relative_path
):

    expected = PUBLIC_HASHES[
        relative_path
    ]

    # Search likely local copies first.
    candidate_roots = [
        PROJECT_ROOT,
        PROJECT_ROOT
        / "GLASS_GANet",
        PROJECT_ROOT
        / "GLASS_GANet/integration",
        Path("/content/GLASS"),
        Path("/content/GLASS_public_validation"),
        Path("/content/glass_public_inspect"),
    ]

    for root in candidate_roots:

        if not root.exists():
            continue

        direct = (
            root
            / relative_path
        )

        candidates = []

        if direct.is_file():
            candidates.append(
                direct
            )

        try:
            candidates.extend(
                root.rglob(
                    Path(
                        relative_path
                    ).name
                )
            )
        except Exception:
            pass

        for candidate in candidates:

            try:
                raw = (
                    candidate
                    .read_bytes()
                )
            except Exception:
                continue

            if (
                bytes_sha256(raw)
                ==
                expected
            ):

                print(
                    "Pinned GLASS source:",
                    candidate
                )

                return raw

    # Exact pinned GitHub fallback.
    url = (
        "https://raw.githubusercontent.com/"
        "BangyaoZhao/GLASS/"
        f"{GLASS_COMMIT}/"
        f"{relative_path}"
    )

    print(
        "Fetching pinned GLASS source:",
        relative_path
    )

    with urllib.request.urlopen(
        url,
        timeout=30
    ) as response:

        raw = response.read()

    require(
        bytes_sha256(raw)
        ==
        expected,
        f"Hash mismatch for {relative_path}"
    )

    return raw


glass_source = obtain_pinned_source(
    "code/glass.py"
).decode("utf-8")

helper_source = obtain_pinned_source(
    "code/helper_funcs.py"
).decode("utf-8")


# Extract ONLY official predict_prob and evaluate_chr_accu.

glass_tree = ast.parse(
    glass_source
)

glass_class = next(
    node
    for node in glass_tree.body
    if (
        isinstance(
            node,
            ast.ClassDef
        )
        and
        node.name == "Glass"
    )
)

predict_node = next(
    node
    for node in glass_class.body
    if (
        isinstance(
            node,
            ast.FunctionDef
        )
        and
        node.name == "predict_prob"
    )
)

helper_tree = ast.parse(
    helper_source
)

evaluate_node = next(
    node
    for node in helper_tree.body
    if (
        isinstance(
            node,
            ast.FunctionDef
        )
        and
        node.name == "evaluate_chr_accu"
    )
)

official_namespace = {
    "np": np,
    "softmax": softmax,
}

exec(
    compile(
        ast.Module(
            body=[
                predict_node,
                evaluate_node
            ],
            type_ignores=[]
        ),
        "official_GLASS_locked_outer",
        "exec"
    ),
    official_namespace
)

official_predict_prob = (
    official_namespace[
        "predict_prob"
    ]
)

official_evaluate_chr_accu = (
    official_namespace[
        "evaluate_chr_accu"
    ]
)

print(
    "Pinned official GLASS prediction/evaluator: PASS"
)


# ============================================================
# 5. PURE GLASS OUTER PREDICTION
# ============================================================

with np.load(
    GLASS_STAGE2,
    allow_pickle=False
) as archive:

    beta_outer = np.asarray(
        archive["betaMat"],
        dtype=np.float32
    )

    glass_fit_seconds = float(
        archive[
            "total_fitting_seconds"
        ]
    )


require(
    beta_outer.shape == (8, 128),
    beta_outer.shape
)

np.testing.assert_allclose(
    beta_outer,
    beta_fold6,
    rtol=0,
    atol=0
)


glass_proxy = types.SimpleNamespace(
    betaMat=beta_outer
)

pure_glass_prob = np.asarray(
    official_predict_prob(
        glass_proxy,
        X_outer[..., 0],
        method="median"
    ),
    dtype=np.float64
)

require(
    pure_glass_prob.shape
    ==
    (100, 6),
    pure_glass_prob.shape
)

require(
    np.isfinite(
        pure_glass_prob
    ).all(),
    "Non-finite GLASS predictions."
)

np.testing.assert_allclose(
    pure_glass_prob.sum(axis=1),
    1.0,
    rtol=1e-5,
    atol=1e-5
)

print("Pure GLASS prediction: PASS")


# ============================================================
# 6. INTERACTION FEATURES FOR OUTER DATA
# ============================================================

print(
    "Computing outer Fisher-z features..."
)

X0_outer = compute_fisher_features(
    X_outer
)

require(
    X0_outer.shape
    ==
    (100, 6, 28),
    X0_outer.shape
)


# ============================================================
# 7. LOAD EACH NEURAL SEED + PREDICT
# ============================================================

structured_seed_probabilities = []
gated_seed_probabilities = []
interaction_seed_probabilities = []
binary_seed_probabilities = []


def load_checkpoint(
    model,
    seed,
    model_name
):

    checkpoint = (
        neural_run_dir(
            seed,
            model_name
        )
        /
        "best.weights.h5"
    )

    require(
        checkpoint.is_file(),
        checkpoint
    )

    model.load_weights(
        checkpoint
    )

    require(
        int(model.count_params())
        ==
        EXPECTED_PARAMETER_COUNTS[
            model_name
        ],
        f"{model_name}: wrong parameter count."
    )

    return model


for seed in SEEDS:

    print(
        f"\nPredicting neural seed {seed}..."
    )


    # --------------------------------------------------------
    # Structured
    # --------------------------------------------------------

    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(seed)

    model = build_structured_dbnet()

    model = load_checkpoint(
        model,
        seed,
        "structured_dbnet"
    )

    p = model.predict(
        X_outer,
        batch_size=32,
        verbose=0
    )

    structured_seed_probabilities.append(
        np.asarray(
            p,
            dtype=np.float64
        )
    )

    del model
    gc.collect()


    # --------------------------------------------------------
    # GLASS-gated
    # --------------------------------------------------------

    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(seed)

    model = build_gated_hybrid(
        beta_fold6
    )

    model = load_checkpoint(
        model,
        seed,
        "glass_gated_dbnet"
    )

    p = model.predict(
        X_outer,
        batch_size=32,
        verbose=0
    )

    gated_seed_probabilities.append(
        np.asarray(
            p,
            dtype=np.float64
        )
    )

    del model
    gc.collect()


    # --------------------------------------------------------
    # Interaction
    # --------------------------------------------------------

    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(seed)

    model = build_interaction_dbnet()

    model = load_checkpoint(
        model,
        seed,
        "interaction_dbnet"
    )

    p = model.predict(
        [
            X_outer,
            X0_outer
        ],
        batch_size=32,
        verbose=0
    )

    interaction_seed_probabilities.append(
        np.asarray(
            p,
            dtype=np.float64
        )
    )

    del model
    gc.collect()


    # --------------------------------------------------------
    # Binary
    # --------------------------------------------------------

    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(seed)

    model = build_binary_dbnet()

    model = load_checkpoint(
        model,
        seed,
        "binary_dbnet"
    )

    X_outer_binary = (
        X_outer
        .reshape(
            600,
            8,
            128,
            1
        )
    )

    p = model.predict(
        X_outer_binary,
        batch_size=64,
        verbose=0
    )

    binary_seed_probabilities.append(
        np.asarray(
            p,
            dtype=np.float64
        )
    )

    del model
    gc.collect()


# ============================================================
# 8. THREE-SEED ENSEMBLES
# ============================================================

structured_seed_probabilities = (
    np.stack(
        structured_seed_probabilities,
        axis=0
    )
)

gated_seed_probabilities = (
    np.stack(
        gated_seed_probabilities,
        axis=0
    )
)

interaction_seed_probabilities = (
    np.stack(
        interaction_seed_probabilities,
        axis=0
    )
)

binary_seed_probabilities = (
    np.stack(
        binary_seed_probabilities,
        axis=0
    )
)


require(
    structured_seed_probabilities.shape
    ==
    (3, 100, 6),
    structured_seed_probabilities.shape
)

require(
    gated_seed_probabilities.shape
    ==
    (3, 100, 6),
    gated_seed_probabilities.shape
)

require(
    interaction_seed_probabilities.shape
    ==
    (3, 100, 6),
    interaction_seed_probabilities.shape
)

require(
    binary_seed_probabilities.shape
    ==
    (3, 600, 2),
    binary_seed_probabilities.shape
)


# Frozen rule:
# AVERAGE PROBABILITIES FIRST.

structured_prob = np.mean(
    structured_seed_probabilities,
    axis=0
)

gated_prob = np.mean(
    gated_seed_probabilities,
    axis=0
)

interaction_prob = np.mean(
    interaction_seed_probabilities,
    axis=0
)

binary_ensemble_2class = np.mean(
    binary_seed_probabilities,
    axis=0
)


# ============================================================
# 9. FROZEN BINARY -> SIX-CHOICE CONVERSION
# ============================================================

EPSILON = 1e-7

binary_target_probability = (
    binary_ensemble_2class[
        :,
        1
    ]
)

binary_target_probability = np.clip(
    binary_target_probability,
    EPSILON,
    1.0 - EPSILON
)

binary_target_logit = np.log(
    binary_target_probability
    /
    (
        1.0
        -
        binary_target_probability
    )
)

binary_choice_logits = (
    binary_target_logit
    .reshape(
        100,
        6
    )
)

# T = 1 exactly.
binary_prob = softmax(
    binary_choice_logits,
    axis=1
)


# ============================================================
# 10. PROBABILITY AUDIT
# ============================================================

all_probability_sets = {
    "Pure GLASS":
        pure_glass_prob,

    "Binary DBNet T=1":
        binary_prob,

    "Structured DBNet":
        structured_prob,

    "GLASS-gated DBNet":
        gated_prob,

    "Interaction DBNet":
        interaction_prob,
}

for model_name, probabilities in (
    all_probability_sets.items()
):

    require(
        probabilities.shape
        ==
        (100, 6),
        (
            model_name,
            probabilities.shape
        )
    )

    require(
        np.isfinite(
            probabilities
        ).all(),
        f"{model_name}: non-finite probabilities."
    )

    require(
        np.all(
            probabilities >= 0
        ),
        f"{model_name}: negative probabilities."
    )

    np.testing.assert_allclose(
        probabilities.sum(axis=1),
        1.0,
        rtol=1e-5,
        atol=1e-5
    )

print(
    "\nAll probability audits: PASS"
)


# ============================================================
# 11. FROZEN METRICS
# ============================================================

def evaluate_six_choice(
    probabilities
):

    probabilities = np.asarray(
        probabilities,
        dtype=np.float64
    )

    clipped = np.clip(
        probabilities,
        1e-8,
        1.0
    )

    nll = float(
        -np.mean(
            np.sum(
                y_outer
                *
                np.log(
                    clipped
                ),
                axis=1
            )
        )
    )

    half_accuracy = float(
        np.mean(
            probabilities.argmax(
                axis=1
            )
            ==
            y_outer.argmax(
                axis=1
            )
        )
    )

    grouped_probabilities = (
        probabilities
        .reshape(
            5,
            10,
            2,
            6
        )
    )

    grouped_labels = (
        y_outer
        .reshape(
            5,
            10,
            2,
            6
        )
    )

    character_curve = np.asarray(
        official_evaluate_chr_accu(
            grouped_probabilities,
            grouped_labels
        ),
        dtype=np.float64
    )

    require(
        character_curve.shape
        ==
        (10,),
        character_curve.shape
    )

    require(
        np.all(
            (
                character_curve >= 0
            )
            &
            (
                character_curve <= 1
            )
        ),
        "Invalid character accuracy."
    )

    return {
        "NLL":
            nll,

        "Half":
            half_accuracy,

        **{
            f"Acc@{sequence}":
                float(
                    character_curve[
                        sequence - 1
                    ]
                )
            for sequence in range(
                1,
                11
            )
        },
    }, character_curve


rows = []
curves = {}

for model_name, probabilities in (
    all_probability_sets.items()
):

    metrics, curve = (
        evaluate_six_choice(
            probabilities
        )
    )

    curves[
        model_name
    ] = curve

    rows.append({
        "Model":
            model_name,

        **metrics
    })


results = pd.DataFrame(
    rows
)


# ============================================================
# 12. BINARY ORDINARY EVENT-LEVEL METRICS
# ============================================================

binary_true = (
    y_outer
    .reshape(-1)
    .astype(np.int32)
)

binary_score = (
    binary_target_probability
)

binary_pred = (
    binary_score
    >= 0.5
).astype(np.int32)

binary_metrics = {
    "ordinary_accuracy":
        float(
            accuracy_score(
                binary_true,
                binary_pred
            )
        ),

    "balanced_accuracy":
        float(
            balanced_accuracy_score(
                binary_true,
                binary_pred
            )
        ),

    "target_precision":
        float(
            precision_score(
                binary_true,
                binary_pred,
                zero_division=0
            )
        ),

    "target_recall":
        float(
            recall_score(
                binary_true,
                binary_pred,
                zero_division=0
            )
        ),

    "target_f1":
        float(
            f1_score(
                binary_true,
                binary_pred,
                zero_division=0
            )
        ),

    "macro_f1":
        float(
            f1_score(
                binary_true,
                binary_pred,
                average="macro",
                zero_division=0
            )
        ),

    "weighted_f1":
        float(
            f1_score(
                binary_true,
                binary_pred,
                average="weighted",
                zero_division=0
            )
        ),

    "target_auc":
        float(
            roc_auc_score(
                binary_true,
                binary_score
            )
        ),
}


# ============================================================
# 13. PRINT LOCKED RESULT
# ============================================================

print(
    "\n"
    +
    "=" * 110
)

print(
    "A01 OUTER FOLD 6 — CONFIRMATORY RESULT"
)

print(
    "=" * 110
)

print(
    results
    .round(4)
    .to_string(
        index=False
    )
)


structured_acc4 = float(
    results.loc[
        results["Model"]
        ==
        "Structured DBNet",
        "Acc@4"
    ].iloc[0]
)

gated_acc4 = float(
    results.loc[
        results["Model"]
        ==
        "GLASS-gated DBNet",
        "Acc@4"
    ].iloc[0]
)


print(
    "\nPRIMARY COMPARISON"
)

print(
    "  Structured Acc@4:",
    f"{structured_acc4:.4f}"
)

print(
    "  GLASS-gated Acc@4:",
    f"{gated_acc4:.4f}"
)

print(
    "  ΔAcc@4 gated-structured:",
    f"{gated_acc4 - structured_acc4:+.4f}"
)


print(
    "\nBinary outer metrics"
)

for key, value in (
    binary_metrics.items()
):

    print(
        f"  {key}: "
        f"{value:.4f}"
    )


print(
    "\nGLASS CPU fitting time:",
    f"{glass_fit_seconds / 60:.2f}",
    "min"
)


# ============================================================
# 14. SAVE LOCKED OUTER PREDICTIONS
# ============================================================

OUTER_RESULT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

temporary_npz = (
    OUTER_RESULT_DIR
    /
    "outer_predictions.tmp"
)

with temporary_npz.open(
    "wb"
) as handle:

    np.savez_compressed(
        handle,

        subject=np.asarray(
            SUBJECT
        ),

        outer_fold=np.asarray(
            FOLD,
            dtype=np.int64
        ),

        outer_characters_1based=np.asarray(
            OUTER_CHARACTERS,
            dtype=np.int64
        ),

        labels=y_outer.astype(
            np.int8
        ),

        pure_glass_probabilities=
            pure_glass_prob.astype(
                np.float32
            ),

        binary_probabilities=
            binary_prob.astype(
                np.float32
            ),

        structured_probabilities=
            structured_prob.astype(
                np.float32
            ),

        glass_gated_probabilities=
            gated_prob.astype(
                np.float32
            ),

        interaction_probabilities=
            interaction_prob.astype(
                np.float32
            ),

        structured_seed_probabilities=
            structured_seed_probabilities.astype(
                np.float32
            ),

        glass_gated_seed_probabilities=
            gated_seed_probabilities.astype(
                np.float32
            ),

        interaction_seed_probabilities=
            interaction_seed_probabilities.astype(
                np.float32
            ),

        binary_seed_probabilities=
            binary_seed_probabilities.astype(
                np.float32
            ),

        binary_ensemble_target_probability=
            binary_target_probability.astype(
                np.float32
            ),

        fisher_features=
            X0_outer.astype(
                np.float32
            ),
    )


temporary_npz.replace(
    OUTER_PREDICTIONS_PATH
)

prediction_sha256 = file_sha256(
    OUTER_PREDICTIONS_PATH
)


# ============================================================
# 15. SAVE RESULT MANIFEST
# ============================================================

manifest = {
    "status":
        "completed",

    "subject":
        SUBJECT,

    "outer_fold":
        FOLD,

    "outer_characters_1based":
        OUTER_CHARACTERS,

    "primary_endpoint":
        "character_accuracy_4",

    "primary_comparison":
        "glass_gated_dbnet_minus_structured_dbnet",

    "seed_ensemble_rule":
        "mean probabilities across seeds 1,2,3 before metrics",

    "binary_conversion_rule":
        (
            "mean 2-class probabilities across seeds; "
            "target p clipped at 1e-7; "
            "logit(p/(1-p)); "
            "six-choice softmax at T=1"
        ),

    "glass_prediction_method":
        "official predict_prob method=median",

    "glass_commit":
        GLASS_COMMIT,

    "neural_seeds":
        SEEDS,

    "outer_test_used_for_training":
        False,

    "outer_test_used_for_model_selection":
        False,

    "outer_test_first_evaluated_after_all_fits":
        True,

    "metrics":
        {
            row["Model"]: {
                key:
                    float(value)
                for key, value in row.items()
                if key != "Model"
            }
            for row in rows
        },

    "binary_metrics":
        binary_metrics,

    "primary_result": {
        "structured_acc4":
            structured_acc4,

        "glass_gated_acc4":
            gated_acc4,

        "delta_acc4":
            gated_acc4
            -
            structured_acc4,
    },

    "glass_cpu_fitting_seconds":
        glass_fit_seconds,

    "outer_predictions":
        str(
            OUTER_PREDICTIONS_PATH
        ),

    "outer_predictions_sha256":
        prediction_sha256,

    "completed_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),
}


temporary_json = (
    OUTER_RESULT_DIR
    /
    "outer_result_manifest.tmp"
)

temporary_json.write_text(
    json.dumps(
        manifest,
        indent=2,
        allow_nan=False
    ),
    encoding="utf-8"
)

temporary_json.replace(
    OUTER_MANIFEST_PATH
)


# ============================================================
# 16. FINAL AUDIT
# ============================================================

require(
    OUTER_PREDICTIONS_PATH.is_file(),
    "Prediction file missing after save."
)

require(
    OUTER_MANIFEST_PATH.is_file(),
    "Manifest missing after save."
)

require(
    file_sha256(
        OUTER_PREDICTIONS_PATH
    )
    ==
    prediction_sha256,
    "Prediction hash changed."
)


print(
    "\nSaved:"
)

print(
    OUTER_PREDICTIONS_PATH
)

print(
    "SHA256:",
    prediction_sha256
)

print(
    "\nIMPORTANT:"
)

print(
    "  Fold 6 is now permanently opened."
)

print(
    "  Do NOT modify model/protocol based on this result."
)

print(
    "\nSTATUS: PASS — A01 Fold 6 completed under frozen protocol."
)

print(
    "\n"
    +
    "=" * 90
)

print(
    "A01 FOLD 6: COMPLETE"
)

print(
    "=" * 90
)
```

```python
# ============================================================
# A01 OUTER FOLD 7 — LOCKED OUTER EVALUATION
#
# NO TRAINING
# NO MODEL SELECTION
# NO TEMPERATURE TUNING
#
# Frozen rules:
#   - average neural probabilities across seeds 1,2,3 FIRST
#   - Binary: average 2-class probabilities first,
#             p -> log(p/(1-p)) -> six-choice softmax T=1
#   - Pure GLASS: official median prediction
#
# Primary endpoint:
#   Acc@4
#
# Primary comparison:
#   GLASS-gated DBNet vs Structured DBNet
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import ast
import gc
import hashlib
import json
import types
import urllib.request

import numpy as np
import pandas as pd
import tensorflow as tf

from scipy.special import softmax

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
)


# ============================================================
# 1. FROZEN FOLD
# ============================================================

SUBJECT = "A01"
FOLD = 7
SEEDS = [1, 2, 3]

OUTER_CHARACTERS = [
    1, 3, 13, 14, 30
]

MODEL_NAMES = [
    "structured_dbnet",
    "glass_gated_dbnet",
    "interaction_dbnet",
    "binary_dbnet",
]

EXPECTED_PARAMETER_COUNTS = {
    "structured_dbnet": 3961,
    "glass_gated_dbnet": 4986,
    "interaction_dbnet": 3989,
    "binary_dbnet": 4090,
}

OUTER_RESULT_DIR = (
    CV_ROOT
    / "outer_results"
    / SUBJECT
    / "outer_fold_7"
)

OUTER_PREDICTIONS_PATH = (
    OUTER_RESULT_DIR
    / "outer_predictions.npz"
)

OUTER_MANIFEST_PATH = (
    OUTER_RESULT_DIR
    / "outer_result_manifest.json"
)


def require(condition, message):
    if not condition:
        raise RuntimeError(message)


def file_sha256(path):
    h = hashlib.sha256()

    with Path(path).open("rb") as f:
        for block in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            h.update(block)

    return h.hexdigest()


def neural_run_dir(seed, model_name):

    return (
        MODEL_ROOT
        / f"seed_{seed}"
        / model_name
    )


# ============================================================
# 2. PRE-OPEN AUDIT
# ============================================================

print("=" * 100)
print("A01 OUTER FOLD 7 — LOCKED EVALUATION")
print("=" * 100)

require(
    not OUTER_PREDICTIONS_PATH.exists(),
    "Fold 7 was already evaluated."
)

require(
    not OUTER_MANIFEST_PATH.exists(),
    "Fold 7 result manifest already exists."
)

for seed in SEEDS:

    for model_name in MODEL_NAMES:

        run_dir = neural_run_dir(
            seed,
            model_name
        )

        require(
            (run_dir / "best.weights.h5").is_file(),
            f"Missing {seed} {model_name} weights"
        )

        require(
            (run_dir / "training_summary.json").is_file(),
            f"Missing {seed} {model_name} summary"
        )

require(
    GLASS_STAGE2.is_file(),
    "Fold-7 GLASS Stage-2 fit missing."
)

print("All 12 neural fits: PASS")
print("Fold-local GLASS fit: PASS")

print("\nOuter performance evaluation starts NOW.")


# ============================================================
# 3. FIRST ACCESS TO OUTER FOLD 7
# ============================================================

outer_idx = (
    np.asarray(
        OUTER_CHARACTERS,
        dtype=np.int64
    )
    - 1
)

X_outer_character = np.asarray(
    X_all[outer_idx],
    dtype=np.float32
)

y_outer_character = np.asarray(
    y_all[outer_idx],
    dtype=np.int8
)

require(
    X_outer_character.shape
    ==
    (5, 10, 2, 6, 8, 128),
    X_outer_character.shape
)

require(
    y_outer_character.shape
    ==
    (5, 10, 2, 6),
    y_outer_character.shape
)


targets = y_outer_character.argmax(
    axis=-1
)

np.testing.assert_array_equal(
    targets,
    np.repeat(
        targets[:, :1, :],
        repeats=10,
        axis=1
    )
)


X_outer = (
    X_outer_character
    .reshape(
        100,
        6,
        8,
        128
    )
    [..., np.newaxis]
)

y_outer = (
    y_outer_character
    .reshape(
        100,
        6
    )
    .astype(np.float32)
)

require(
    X_outer.shape
    ==
    (100, 6, 8, 128, 1),
    X_outer.shape
)

require(
    y_outer.shape
    ==
    (100, 6),
    y_outer.shape
)

print("Outer EEG:", X_outer.shape)
print("Outer labels:", y_outer.shape)


# ============================================================
# 4. PINNED OFFICIAL GLASS EVALUATOR
# ============================================================

GLASS_COMMIT = (
    "db32c2abd9211dec225a7f39a33963f8a0b21d7f"
)

PUBLIC_HASHES = {
    "code/glass.py":
        "00de59d0ba107a7a5770565bdfb5bf7dfbb372f92bed2c4aecd1bb21deda652d",

    "code/helper_funcs.py":
        "f9cf4a5134acbbda7fcb55e6503096f1698b0f9e63e16a2a4f9252dd51eb54d9",
}


def bytes_sha256(raw):
    return hashlib.sha256(raw).hexdigest()


def obtain_pinned_source(relative_path):

    expected = PUBLIC_HASHES[
        relative_path
    ]

    roots = [
        PROJECT_ROOT,
        PROJECT_ROOT / "GLASS_GANet",
        PROJECT_ROOT / "GLASS_GANet/integration",
        Path("/content/GLASS"),
    ]

    for root in roots:

        if not root.exists():
            continue

        try:
            candidates = list(
                root.rglob(
                    Path(relative_path).name
                )
            )
        except Exception:
            candidates = []

        for candidate in candidates:

            try:
                raw = candidate.read_bytes()
            except Exception:
                continue

            if bytes_sha256(raw) == expected:

                print(
                    "Pinned GLASS source:",
                    candidate
                )

                return raw

    url = (
        "https://raw.githubusercontent.com/"
        "BangyaoZhao/GLASS/"
        f"{GLASS_COMMIT}/"
        f"{relative_path}"
    )

    with urllib.request.urlopen(
        url,
        timeout=30
    ) as response:

        raw = response.read()

    require(
        bytes_sha256(raw) == expected,
        f"GLASS hash mismatch: {relative_path}"
    )

    return raw


glass_source = obtain_pinned_source(
    "code/glass.py"
).decode("utf-8")

helper_source = obtain_pinned_source(
    "code/helper_funcs.py"
).decode("utf-8")


glass_tree = ast.parse(
    glass_source
)

glass_class = next(
    node
    for node in glass_tree.body
    if (
        isinstance(node, ast.ClassDef)
        and node.name == "Glass"
    )
)

predict_node = next(
    node
    for node in glass_class.body
    if (
        isinstance(node, ast.FunctionDef)
        and node.name == "predict_prob"
    )
)


helper_tree = ast.parse(
    helper_source
)

evaluate_node = next(
    node
    for node in helper_tree.body
    if (
        isinstance(node, ast.FunctionDef)
        and node.name == "evaluate_chr_accu"
    )
)


official_namespace = {
    "np": np,
    "softmax": softmax,
}

exec(
    compile(
        ast.Module(
            body=[
                predict_node,
                evaluate_node
            ],
            type_ignores=[]
        ),
        "official_glass_fold7",
        "exec"
    ),
    official_namespace
)

official_predict_prob = (
    official_namespace[
        "predict_prob"
    ]
)

official_evaluate_chr_accu = (
    official_namespace[
        "evaluate_chr_accu"
    ]
)

print(
    "Pinned official GLASS prediction/evaluator: PASS"
)


# ============================================================
# 5. PURE GLASS
# ============================================================

with np.load(
    GLASS_STAGE2,
    allow_pickle=False
) as archive:

    beta_outer = np.asarray(
        archive["betaMat"],
        dtype=np.float32
    )

    glass_fit_seconds = float(
        archive[
            "total_fitting_seconds"
        ]
    )


np.testing.assert_allclose(
    beta_outer,
    beta_fold7,
    rtol=0,
    atol=0
)

glass_proxy = types.SimpleNamespace(
    betaMat=beta_outer
)

pure_glass_prob = np.asarray(
    official_predict_prob(
        glass_proxy,
        X_outer[..., 0],
        method="median"
    ),
    dtype=np.float64
)

require(
    pure_glass_prob.shape == (100, 6),
    pure_glass_prob.shape
)

print("Pure GLASS prediction: PASS")


# ============================================================
# 6. OUTER FISHER FEATURES
# ============================================================

print(
    "Computing outer Fisher-z features..."
)

X0_outer = compute_fisher_features(
    X_outer
)

require(
    X0_outer.shape == (100, 6, 28),
    X0_outer.shape
)


# ============================================================
# 7. NEURAL PREDICTIONS
# ============================================================

structured_seed_probabilities = []
gated_seed_probabilities = []
interaction_seed_probabilities = []
binary_seed_probabilities = []

X_outer_binary = (
    X_outer
    .reshape(
        600,
        8,
        128,
        1
    )
)


def load_checkpoint(
    model,
    seed,
    model_name
):

    checkpoint = (
        neural_run_dir(
            seed,
            model_name
        )
        /
        "best.weights.h5"
    )

    model.load_weights(
        checkpoint
    )

    require(
        int(model.count_params())
        ==
        EXPECTED_PARAMETER_COUNTS[
            model_name
        ],
        f"{model_name}: parameter mismatch"
    )

    return model


for seed in SEEDS:

    print(
        f"\nPredicting neural seed {seed}..."
    )


    # Structured
    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(seed)

    model = build_structured_dbnet()

    load_checkpoint(
        model,
        seed,
        "structured_dbnet"
    )

    structured_seed_probabilities.append(
        np.asarray(
            model.predict(
                X_outer,
                batch_size=32,
                verbose=0
            ),
            dtype=np.float64
        )
    )

    del model
    gc.collect()


    # Gated
    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(seed)

    model = build_gated_hybrid(
        beta_fold7
    )

    load_checkpoint(
        model,
        seed,
        "glass_gated_dbnet"
    )

    gated_seed_probabilities.append(
        np.asarray(
            model.predict(
                X_outer,
                batch_size=32,
                verbose=0
            ),
            dtype=np.float64
        )
    )

    del model
    gc.collect()


    # Interaction
    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(seed)

    model = build_interaction_dbnet()

    load_checkpoint(
        model,
        seed,
        "interaction_dbnet"
    )

    interaction_seed_probabilities.append(
        np.asarray(
            model.predict(
                [
                    X_outer,
                    X0_outer
                ],
                batch_size=32,
                verbose=0
            ),
            dtype=np.float64
        )
    )

    del model
    gc.collect()


    # Binary
    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(seed)

    model = build_binary_dbnet()

    load_checkpoint(
        model,
        seed,
        "binary_dbnet"
    )

    binary_seed_probabilities.append(
        np.asarray(
            model.predict(
                X_outer_binary,
                batch_size=64,
                verbose=0
            ),
            dtype=np.float64
        )
    )

    del model
    gc.collect()


# ============================================================
# 8. ENSEMBLE
# ============================================================

structured_seed_probabilities = np.stack(
    structured_seed_probabilities,
    axis=0
)

gated_seed_probabilities = np.stack(
    gated_seed_probabilities,
    axis=0
)

interaction_seed_probabilities = np.stack(
    interaction_seed_probabilities,
    axis=0
)

binary_seed_probabilities = np.stack(
    binary_seed_probabilities,
    axis=0
)


structured_prob = np.mean(
    structured_seed_probabilities,
    axis=0
)

gated_prob = np.mean(
    gated_seed_probabilities,
    axis=0
)

interaction_prob = np.mean(
    interaction_seed_probabilities,
    axis=0
)

binary_ensemble_2class = np.mean(
    binary_seed_probabilities,
    axis=0
)


# ============================================================
# 9. BINARY -> SIX-CHOICE, T=1
# ============================================================

EPSILON = 1e-7

binary_target_probability = np.clip(
    binary_ensemble_2class[:, 1],
    EPSILON,
    1.0 - EPSILON
)

binary_target_logit = np.log(
    binary_target_probability
    /
    (
        1.0
        -
        binary_target_probability
    )
)

binary_choice_logits = (
    binary_target_logit
    .reshape(
        100,
        6
    )
)

binary_prob = softmax(
    binary_choice_logits,
    axis=1
)


# ============================================================
# 10. AUDIT
# ============================================================

all_probability_sets = {
    "Pure GLASS":
        pure_glass_prob,

    "Binary DBNet T=1":
        binary_prob,

    "Structured DBNet":
        structured_prob,

    "GLASS-gated DBNet":
        gated_prob,

    "Interaction DBNet":
        interaction_prob,
}

for name, probabilities in (
    all_probability_sets.items()
):

    require(
        probabilities.shape == (100, 6),
        (name, probabilities.shape)
    )

    require(
        np.isfinite(
            probabilities
        ).all(),
        f"{name}: non-finite predictions"
    )

    np.testing.assert_allclose(
        probabilities.sum(axis=1),
        1.0,
        rtol=1e-5,
        atol=1e-5
    )

print(
    "\nAll probability audits: PASS"
)


# ============================================================
# 11. METRICS
# ============================================================

def evaluate_six_choice(
    probabilities
):

    clipped = np.clip(
        probabilities,
        1e-8,
        1.0
    )

    nll = float(
        -np.mean(
            np.sum(
                y_outer
                *
                np.log(clipped),
                axis=1
            )
        )
    )

    half_accuracy = float(
        np.mean(
            probabilities.argmax(axis=1)
            ==
            y_outer.argmax(axis=1)
        )
    )

    grouped_prob = (
        probabilities
        .reshape(
            5,
            10,
            2,
            6
        )
    )

    grouped_labels = (
        y_outer
        .reshape(
            5,
            10,
            2,
            6
        )
    )

    curve = np.asarray(
        official_evaluate_chr_accu(
            grouped_prob,
            grouped_labels
        ),
        dtype=np.float64
    )

    require(
        curve.shape == (10,),
        curve.shape
    )

    metrics = {
        "NLL":
            nll,

        "Half":
            half_accuracy,
    }

    for sequence in range(
        1,
        11
    ):

        metrics[
            f"Acc@{sequence}"
        ] = float(
            curve[
                sequence - 1
            ]
        )

    return metrics, curve


rows = []
curves = {}

for name, probabilities in (
    all_probability_sets.items()
):

    metrics, curve = evaluate_six_choice(
        probabilities
    )

    curves[name] = curve

    rows.append(
        {
            "Model":
                name,
            **metrics
        }
    )


results = pd.DataFrame(
    rows
)


# ============================================================
# 12. BINARY EVENT METRICS
# ============================================================

binary_true = (
    y_outer
    .reshape(-1)
    .astype(np.int32)
)

binary_score = (
    binary_target_probability
)

binary_pred = (
    binary_score
    >= 0.5
).astype(np.int32)

binary_metrics = {
    "ordinary_accuracy":
        float(
            accuracy_score(
                binary_true,
                binary_pred
            )
        ),

    "balanced_accuracy":
        float(
            balanced_accuracy_score(
                binary_true,
                binary_pred
            )
        ),

    "target_precision":
        float(
            precision_score(
                binary_true,
                binary_pred,
                zero_division=0
            )
        ),

    "target_recall":
        float(
            recall_score(
                binary_true,
                binary_pred,
                zero_division=0
            )
        ),

    "target_f1":
        float(
            f1_score(
                binary_true,
                binary_pred,
                zero_division=0
            )
        ),

    "macro_f1":
        float(
            f1_score(
                binary_true,
                binary_pred,
                average="macro",
                zero_division=0
            )
        ),

    "weighted_f1":
        float(
            f1_score(
                binary_true,
                binary_pred,
                average="weighted",
                zero_division=0
            )
        ),

    "target_auc":
        float(
            roc_auc_score(
                binary_true,
                binary_score
            )
        ),
}


# ============================================================
# 13. PRINT RESULT
# ============================================================

print(
    "\n"
    +
    "=" * 110
)

print(
    "A01 OUTER FOLD 7 — CONFIRMATORY RESULT"
)

print(
    "=" * 110
)

print(
    results
    .round(4)
    .to_string(
        index=False
    )
)


structured_acc4 = float(
    results.loc[
        results["Model"]
        ==
        "Structured DBNet",
        "Acc@4"
    ].iloc[0]
)

gated_acc4 = float(
    results.loc[
        results["Model"]
        ==
        "GLASS-gated DBNet",
        "Acc@4"
    ].iloc[0]
)


print(
    "\nPRIMARY COMPARISON"
)

print(
    "  Structured Acc@4:",
    f"{structured_acc4:.4f}"
)

print(
    "  GLASS-gated Acc@4:",
    f"{gated_acc4:.4f}"
)

print(
    "  ΔAcc@4 gated-structured:",
    f"{gated_acc4 - structured_acc4:+.4f}"
)


print(
    "\nBinary outer metrics"
)

for key, value in (
    binary_metrics.items()
):

    print(
        f"  {key}: {value:.4f}"
    )


print(
    "\nGLASS CPU fitting time:",
    f"{glass_fit_seconds/60:.2f}",
    "min"
)


# ============================================================
# 14. SAVE PREDICTIONS
# ============================================================

OUTER_RESULT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

temporary_npz = (
    OUTER_RESULT_DIR
    /
    "outer_predictions.tmp"
)

with temporary_npz.open(
    "wb"
) as handle:

    np.savez_compressed(
        handle,

        subject=np.asarray(
            SUBJECT
        ),

        outer_fold=np.asarray(
            FOLD,
            dtype=np.int64
        ),

        outer_characters_1based=np.asarray(
            OUTER_CHARACTERS,
            dtype=np.int64
        ),

        labels=y_outer.astype(
            np.int8
        ),

        pure_glass_probabilities=
            pure_glass_prob.astype(
                np.float32
            ),

        binary_probabilities=
            binary_prob.astype(
                np.float32
            ),

        structured_probabilities=
            structured_prob.astype(
                np.float32
            ),

        glass_gated_probabilities=
            gated_prob.astype(
                np.float32
            ),

        interaction_probabilities=
            interaction_prob.astype(
                np.float32
            ),

        structured_seed_probabilities=
            structured_seed_probabilities.astype(
                np.float32
            ),

        glass_gated_seed_probabilities=
            gated_seed_probabilities.astype(
                np.float32
            ),

        interaction_seed_probabilities=
            interaction_seed_probabilities.astype(
                np.float32
            ),

        binary_seed_probabilities=
            binary_seed_probabilities.astype(
                np.float32
            ),

        binary_ensemble_target_probability=
            binary_target_probability.astype(
                np.float32
            ),

        fisher_features=
            X0_outer.astype(
                np.float32
            ),
    )


temporary_npz.replace(
    OUTER_PREDICTIONS_PATH
)

prediction_sha256 = file_sha256(
    OUTER_PREDICTIONS_PATH
)


# ============================================================
# 15. SAVE MANIFEST
# ============================================================

manifest = {
    "status":
        "completed",

    "subject":
        SUBJECT,

    "outer_fold":
        FOLD,

    "outer_characters_1based":
        OUTER_CHARACTERS,

    "primary_endpoint":
        "character_accuracy_4",

    "primary_comparison":
        "glass_gated_dbnet_minus_structured_dbnet",

    "seed_ensemble_rule":
        "mean probabilities across seeds 1,2,3 before metrics",

    "binary_conversion_rule":
        (
            "mean 2-class probabilities across seeds; "
            "target p clipped at 1e-7; "
            "logit(p/(1-p)); "
            "six-choice softmax at T=1"
        ),

    "glass_prediction_method":
        "official predict_prob method=median",

    "glass_commit":
        GLASS_COMMIT,

    "neural_seeds":
        SEEDS,

    "outer_test_used_for_training":
        False,

    "outer_test_used_for_model_selection":
        False,

    "outer_test_first_evaluated_after_all_fits":
        True,

    "metrics": {
        row["Model"]: {
            key:
                float(value)
            for key, value in row.items()
            if key != "Model"
        }
        for row in rows
    },

    "binary_metrics":
        binary_metrics,

    "primary_result": {
        "structured_acc4":
            structured_acc4,

        "glass_gated_acc4":
            gated_acc4,

        "delta_acc4":
            gated_acc4
            -
            structured_acc4,
    },

    "glass_cpu_fitting_seconds":
        glass_fit_seconds,

    "outer_predictions":
        str(
            OUTER_PREDICTIONS_PATH
        ),

    "outer_predictions_sha256":
        prediction_sha256,

    "completed_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),
}


temporary_json = (
    OUTER_RESULT_DIR
    /
    "outer_result_manifest.tmp"
)

temporary_json.write_text(
    json.dumps(
        manifest,
        indent=2,
        allow_nan=False
    ),
    encoding="utf-8"
)

temporary_json.replace(
    OUTER_MANIFEST_PATH
)


# ============================================================
# 16. FINAL AUDIT
# ============================================================

require(
    OUTER_PREDICTIONS_PATH.is_file(),
    "Prediction save failed."
)

require(
    OUTER_MANIFEST_PATH.is_file(),
    "Manifest save failed."
)

require(
    file_sha256(
        OUTER_PREDICTIONS_PATH
    )
    ==
    prediction_sha256,
    "Prediction hash changed."
)


print(
    "\nSaved:"
)

print(
    OUTER_PREDICTIONS_PATH
)

print(
    "SHA256:",
    prediction_sha256
)

print(
    "\nIMPORTANT:"
)

print(
    "  Fold 7 is now permanently opened."
)

print(
    "  Do NOT modify model/protocol based on this result."
)

print(
    "\nSTATUS: PASS — A01 Fold 7 completed under frozen protocol."
)

print(
    "\n"
    +
    "=" * 90
)

print(
    "A01 FOLD 7: COMPLETE"
)

print(
    "=" * 90
)
```
